```python
def triangular(x, left, center, right):
    if x <= left:
        return 0

    if x <= center:
        return (x - left) / (center - left)

    if x < right:
        return (right - x) / (right - center)

    return 0
```

```python
warm = triangular(
    x=24,
    left=20,
    center=25,
    right=30
)

print(warm)
```

```python
def fuzzify_temperature(temp):

    cold = triangular(temp, 15, 20, 25)
    warm = triangular(temp, 20, 25, 30)
    hot = triangular(temp, 25, 30, 35)

    return cold, warm, hot
```

```python
print(fuzzify_temperature(24))
```
